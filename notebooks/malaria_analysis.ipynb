```python
# 5. Data Cleaning

Data cleaning is a critical step in the data analysis process. In this section, we identify and address data quality issues such as missing values, duplicate records, inconsistent data types, and formatting problems to ensure that the dataset is accurate, consistent, and ready for analysis.
```

```python
# 7. Data Visualization

Data visualization is a fundamental component of epidemiological data analysis. In this section, various charts are used to explore malaria patterns, compare countries and WHO regions, and examine temporal trends. These visualizations help communicate findings clearly and support evidence-based public health decision-making.
```

```python
## 7.1 Top 10 Countries by Malaria Burden

This bar chart displays the ten countries with the highest reported malaria burden based on the numerical values available in the dataset.
```

```python
top10 = (
    df_clean.groupby("Country")["Numeric_Value"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(12,6))
top10.plot(kind="bar")

plt.title("Top 10 Countries by Malaria Burden")
plt.xlabel("Country")
plt.ylabel("Total Malaria Burden")
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
## 7.2 Malaria Trend Over Time

This line chart illustrates changes in the malaria burden over time using annual aggregated values.
```

```python
trend = (
    df_clean.groupby("Year")["Numeric_Value"]
    .sum()
    .reset_index()
)

plt.figure(figsize=(12,6))

plt.plot(
    trend["Year"],
    trend["Numeric_Value"],
    marker="o"
)

plt.title("Global Malaria Trend Over Time")
plt.xlabel("Year")
plt.ylabel("Total Malaria Burden")

plt.grid(True)

plt.show()
```

```python
## 7.3 Comparison of WHO Regions

This bar chart compares the malaria burden across WHO regions.
```

```python
region = (
    df_clean.groupby("WHO_Region")["Numeric_Value"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10,6))

region.plot(kind="bar")

plt.title("Malaria Burden by WHO Region")
plt.xlabel("WHO Region")
plt.ylabel("Total Malaria Burden")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

```python
## 7.4 Distribution of Malaria Values

This histogram illustrates the distribution of malaria burden values across all observations.
```

```python
plt.figure(figsize=(10,6))

plt.hist(df_clean["Numeric_Value"], bins=30)

plt.title("Distribution of Malaria Burden")
plt.xlabel("Numeric Value")
plt.ylabel("Frequency")

plt.show()
```

```python
## 7.5 Box Plot

The box plot is used to identify the spread of malaria burden values and detect potential outliers.
```

```python
plt.figure(figsize=(10,5))

plt.boxplot(df_clean["Numeric_Value"])

plt.title("Box Plot of Malaria Burden")

plt.show()
```
